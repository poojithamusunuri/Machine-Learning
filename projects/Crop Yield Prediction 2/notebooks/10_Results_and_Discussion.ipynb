# 10 — Results & Discussion

This notebook converts the measured results into project-level conclusions. It distinguishes validation performance from the final test performance and documents limitations.

```python
from pathlib import Path
import pandas as pd
PROJECT_ROOT=Path.cwd(); OUTPUT_DIR=PROJECT_ROOT/"outputs"
cv=pd.read_csv(OUTPUT_DIR/"cross_validation_results.csv"); test=pd.read_csv(OUTPUT_DIR/"model_evaluation_results.csv"); comp=pd.read_csv(OUTPUT_DIR/"model_comparison.csv")
best=comp.sort_values("CV_RMSE_Mean").iloc[0]
print("SELECTED MODEL:",best["Model"])
print("CV RMSE:",round(best["CV_RMSE_Mean"],4),"±",round(best["CV_RMSE_STD"],4))
print("Test MAE:",round(best["MAE"],4))
print("Test RMSE:",round(best["RMSE"],4))
print("Test R²:",round(best["R2"],4))
```

## Discussion

1. The problem is a regression task with a highly skewed target.
2. `Production` is excluded because it can leak information about yield.
3. Missing predictors are imputed inside fitted preprocessing pipelines, while the missing target row is removed.
4. Feature engineering adds fertilizer-per-area and pesticide-per-area using only predictors.
5. Model selection uses cross-validation on the training data.
6. The final test set is kept separate and is reported as an honest generalization check.
7. The tuned XGBoost model is selected because it achieved the lowest mean CV RMSE in the controlled comparison.
8. SVR is included as a benchmark, but its RBF experiment uses a fixed 5,000-row training subset because full-scale RBF SVR is computationally quadratic.
9. Extreme target values make RMSE substantially larger than MAE; this is expected because RMSE heavily penalizes large errors.

```python
discussion=pd.DataFrame({"Item":["Problem type","Leakage control","Validation","Selection rule","Final selected model","SVR limitation"],"Result":["Regression","Production excluded; preprocessing fitted on training data","3-fold CV on training split","Lowest mean CV RMSE",""+str(best["Model"]),"RBF SVR trained on fixed 5,000-row subset"]})
display(discussion)
discussion.to_csv(OUTPUT_DIR/"project_results_discussion.csv",index=False)
```
