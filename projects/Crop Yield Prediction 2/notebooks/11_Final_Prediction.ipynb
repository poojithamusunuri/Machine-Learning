# 11 — Final Prediction / Demonstration

This notebook loads the **selected tuned XGBoost pipeline** and predicts yield for a new, unseen input. The pipeline contains its own preprocessing, so the new sample is transformed exactly as during training.

```python
from pathlib import Path
import pandas as pd, joblib, numpy as np
PROJECT_ROOT=Path.cwd(); MODEL_DIR=PROJECT_ROOT/"models"; OUTPUT_DIR=PROJECT_ROOT/"outputs"
model=joblib.load(MODEL_DIR/"tuned_xgboost_pipeline.pkl")
print("Selected model pipeline loaded successfully:",type(model).__name__)
```

```python
sample_input=pd.DataFrame([{
    "Crop":"Rice", "Crop_Year":2020, "Season":"Kharif", "State":"Andhra Pradesh",
    "Area":1000.0, "Annual_Rainfall":1200.0, "Fertilizer":150000.0, "Pesticide":500.0,
    "Avg_Temperature":25.0, "Max_Temperature":34.0, "Min_Temperature":18.0
}])
sample_input["Fertilizer_per_Area"]=sample_input["Fertilizer"]/(sample_input["Area"]+1e-9)
sample_input["Pesticide_per_Area"]=sample_input["Pesticide"]/(sample_input["Area"]+1e-9)
display(sample_input)
prediction=float(model.predict(sample_input)[0])
print(f"Predicted crop yield: {prediction:.6f}")
print("Units are the same as the dataset's Yield target.")
```

```python
# Sanity checks for the demonstration input.
required=set(model.feature_names_in_) if hasattr(model,"feature_names_in_") else set(sample_input.columns)
print("Prediction completed successfully.")
print("Target column supplied to model?", "Yield" in sample_input.columns)
print("Production supplied to model?", "Production" in sample_input.columns)
```
